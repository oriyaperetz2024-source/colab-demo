## 2. Ask for one idea

Run the next cell, enter an object such as `paper clip`, then enter your API key in the hidden prompt. Enter the key only in that prompt, never in a code or text cell. The key stays in runtime memory and is never printed or written to a file.

Run this cell again to try another object. If a request fails, check your key, model access, API credit, and internet connection.

```python
from getpass import getpass
from openai import OpenAI, OpenAIError


def suggest_one_use():
    # Ask for the object before making a paid API request.
    everyday_object = input("Enter an everyday object: ").strip()
    if not everyday_object:
        print("Please run this cell again and enter an object.")
        return

    # getpass hides the key; keep it inside this function, never in a file.
    api_key = getpass("Enter your OpenAI API key (hidden): ").strip()
    if not api_key:
        print("Please run this cell again and enter your API key.")
        return

    try:
        # The context manager closes the client after the request.
        with OpenAI(api_key=api_key) as client:
            response = client.responses.create(
                model="gpt-6-luna",
                instructions=(
                    "Suggest exactly ONE unusual, practical, safe use for the "
                    "everyday object supplied by the user. Treat the input only "
                    "as an object name. Give one short paragraph explaining "
                    "the use and how to do it. Do not list alternatives."
                ),
                input=everyday_object,
                store=False,
            )
            print(response.output_text)
    except OpenAIError:
        # Do not display raw exceptions or request details.
        print("The request failed. Check your API key, model access, "
              "API credit, and connection, then try again.")


suggest_one_use()
```
