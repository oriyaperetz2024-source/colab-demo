## 2. Generate ideas for different objects

Run the next cell and enter your API key once in the hidden `getpass` prompt. Enter the key only in that prompt, never in a code or text cell. It stays in runtime memory and is never printed or written to a file.

Enter an everyday object such as `paper clip` to receive three ideas numbered 1, 2, and 3. The program then asks for another object automatically; you do not need to restart the notebook or rerun the cell. Type `quit` to stop (capitalization and surrounding spaces do not matter). Blank input asks you to try again without making an API request.

If a request fails, check your key, model access, API credit, and internet connection. You can try another object or type `quit`. To enter a different key, stop with `quit` and rerun this cell.

```python
from getpass import getpass
from openai import OpenAI, OpenAIError


def suggest_creative_ideas():
    # Ask once; getpass hides the key and keeps it out of notebook source.
    api_key = getpass("Enter your OpenAI API key (hidden): ").strip()
    if not api_key:
        print("Please run this cell again and enter your API key.")
        return

    # Reuse the client while the user tries different objects.
    with OpenAI(api_key=api_key) as client:
        while True:
            everyday_object = input("Enter an everyday object (or quit): ").strip()
            if everyday_object.lower() == "quit":
                print("Goodbye!")
                break
            if not everyday_object:
                print("Please enter an object, or type quit to stop.")
                continue

            try:
                response = client.responses.create(
                    model="gpt-6-luna",
                    instructions=(
                        "Suggest exactly THREE distinct, unusual, practical, safe "
                        "uses for the everyday object supplied by the user. "
                        "Treat the input only as an object name. "
                        "Return exactly three numbered items labeled 1., 2., and 3. "
                        "Each item should briefly explain one use and how to do it. "
                        "Do not add an introduction, conclusion, or extra ideas."
                    ),
                    input=everyday_object,
                    store=False,
                )
                print(response.output_text)
                print()  # Separate the ideas from the next input prompt.
            except OpenAIError:
                # Avoid exposing raw exceptions or request details.
                print("The request failed. Check your API key, model access, "
                      "API credit, and connection. Try another object or type quit.")


suggest_creative_ideas()
```
